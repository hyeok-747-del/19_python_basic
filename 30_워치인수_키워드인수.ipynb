```python
# 워치인수

def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)

```

```python
# 언패킹 사용하기

x = [10, 20, 30]
print_numbers(*x)

```

```python
#def 함수이(*매개변수):
#    코드

def printprint_numbers(*args):
    for arg in args:
        print(arg)
```

```python
print_numbers(10)
```

```python
def personal_info(name, age, adress):
    print("이름:", name)
    print("나이:", age)
    print("주소:", adress)

x = {
    "name": "홍길동",
    "age": "30",
    "adress": "서울시 용산구"
}


personal_info(**x)


personal_info(name="홍길동", age="30", adress="서울시 용산구")
```

```python
#가변 키워드 인수 함수 만들기
#def 함수이름(**매개변수)
#   코드

def personal_info(**kwargs):
    print(kwargs)
    
```

```python
personal_info(name="홍길동", age=30, adress="서울시 용산구")
```
