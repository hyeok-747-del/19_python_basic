```python
#14

students = [
    {"이름": "김민수", "국어": 85, "영어": 90, "수학": 78},
    {"이름": "이영희", "국어": 92, "영어": 88, "수학": 95},
    {"이름": "박철수", "국어": 78, "영어": 85, "수학": 82}
]

print("=== 학생 성적 관리 ===")

for s in students:
    s["총점"] = s["국어"] + s["영어"] + s["수학"]
    s["평균"] = s["총점"] / 3

    print("이름:", s["이름"])
    print(f"국어: {s['국어']}점, 영어: {s['영어']}점, 수학: {s['수학']}점")
    print(f"총점: {s['총점']}점, 평균: {s['평균']:.2f}점")
    print("---")
```

```python
#지뢰찾기


# 지뢰찾기

col, row = map(int, input().split())

matrix = []
for i in range(row):
    matrix.append(list(input()))

for i in range(row):
    for j in range(col):
        if matrix[i][j] == '*':
            print('*', end='')
        else:
            count = 0

            for y in range(i-1, i+2):
                for x in range(j-1, j+2):
                    if 0 <= y < row and 0 <= x < col:
                        if matrix[y][x] == '*':
                            count += 1

            print(count, end='')
    print()


```
