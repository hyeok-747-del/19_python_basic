```python
# 덧셈 함수 만들기

def add(a, b):
    print(a+b)

y = add(10,20)
print(y)
```

```python
# 함수의 결과를 반환하기
#def 함수이름(매개변수):
#    return 반환값

def add(a,b):
    return a+b
```

```python
x = add(10, 20)
print(x)
```

```python
# 함수에서 값을 여러개 반환하기
def add_sub(a,b):
    return a+b, a-b

result = add_sub(10, 20)
print (result)
```
