```python
# while 에서 break로 반복문 끝내기

i=0
while True:
    print(i)
    i += 1
    if i == 100:
        break

```

```python
# for 에서 break 반복문 끝내기

for i in range(100):
    print(i)
    if i == 10:
        break
```

```python
# continue 로 코드 실행 건너뛰기

for i in range(100):
    if i % 2 == 0:
        continue
    print(i)
```

```python
i = 0
while i < 100:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
```
